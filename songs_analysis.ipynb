**Importing the pandas library from python

```python
import pandas as pd
df=pd.read_csv("data.csv")
df
```

Import numpy from python and convert the year column into an array to perform array operations on the column using to_numpy() method and print the data.

```python
import numpy as np
year = df["year"].to_numpy()
print(year)
# print(year.dtype)
```

remove null or Nan values from the array using .isnan(array_name) and change type into int using .astype(dataType)

```python
year = year[~np.isnan(year)]
year=year.astype(int)
```

To print the array without any duplicate values we use .unique(array_name) method

```python
unique_year = np.unique(year)

print(unique_year)
```

Count No.of nan values in the array using isnan(array_name).sum()

```python
print(int(np.isnan(year).sum()))
```

Get Minimum value in the array using "np.min(array_name)"

```python
print(int(np.min(year)))
```

Get Maximum value in the array using "np.max(array_name)"

```python
print(int(np.max(year)))
```

Sort the value in ascending order using "np.sort(array_name)"

```python
print(np.sort(year))
```

Find the mean of the values using "np.mean(array_name)"

```python
print(int(np.mean(year)))
```

Find median of the array using "np.median(array_name)"

```python
print(int(np.median(year)))
```

Find range of the array by subtracting min value from max value

```python
range=np.max(unique_year)-np.min(unique_year)
print(range)
```

Removing rows which consists of whose year is not mentioned.

```python
df = df.dropna(subset=["year"])
```

Change artists column into artist_name .

```python
df = df.rename(columns={"artists": "artist_name"})
df.info()
```

Import matplotlib to view data in the form of graph.

```python
import matplotlib.pyplot as plt

songs_per_year = df["year"].value_counts().sort_index()

plt.plot(songs_per_year.index, songs_per_year.values)

plt.title("Number of Songs Released Each Year")
plt.xlabel("Year")
plt.ylabel("Number of Songs")
plt.show()
```

To discover top 10 artists by no.of songs they released .

```python
top_artists = df["artist_name"].value_counts().head(10)

plt.bar(top_artists.index, top_artists.values)

plt.title("Top 10 Artists by Number of Songs")
plt.xlabel("Artist")
plt.ylabel("Number of Songs")
plt.xticks(rotation=45)
plt.show()
```

Histogram of the no.of songs released in each year.

```python
plt.hist(df["year"].dropna(), bins=20)

plt.title("Distribution of Song Years")
plt.xlabel("Year")
plt.ylabel("Number of Songs")
plt.show()
```

To get pie chart of the top 5 artists of with percentage.

```python
top_5_artists = df["artist_name"].value_counts().head(5)

plt.pie(
    top_5_artists.values,
    labels=top_5_artists.index,
    autopct="%1.1f%%"
)

plt.title("Percentage of Songs from Top 5 Artists")
plt.show()
```

To plot the time of the each song on the graph iin each year.

```python
import matplotlib.pyplot as plt

plt.scatter(df["year"], df["duration_ms"])

plt.title("Year vs Duration")
plt.xlabel("Year")
plt.ylabel("Duration (ms)")

plt.show()
```

To graph the popularity of songs the in each year.

```python
average_popularity = df.groupby("year")["popularity"].mean()

plt.plot(average_popularity.index, average_popularity.values)

plt.title("Average Popularity by Year")
plt.xlabel("Year")
plt.ylabel("Average Popularity")
plt.show()
```
